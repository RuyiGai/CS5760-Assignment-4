# Q1: NLP Preprocessing Pipeline

```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

def preprocess_sentence(sentence):
    # 1. Split the sentence into words and punctuation.
    tokens = re.findall(r"\w+|[^\w\s]", sentence)
    print("Original Tokens:", tokens)

    # 2. Remove English stopwords and punctuation.
    english_stopwords = set(stopwords.words("english"))
    filtered_tokens = [
        token
        for token in tokens
        if token.isalpha() and token.lower() not in english_stopwords
    ]
    print("Tokens Without Stopwords:", filtered_tokens)

    # 3. Convert the remaining words to lowercase and stem them.
    stemmer = PorterStemmer()
    stemmed_words = [stemmer.stem(token.lower()) for token in filtered_tokens]
    print("Stemmed Words:", stemmed_words)

if __name__ == "__main__":
    try:
        stopwords.words("english")
    except LookupError:
        nltk.download("stopwords", quiet=True)

    sentence = "NLP techniques are used in virtual assistants like Alexa and Siri."
    preprocess_sentence(sentence)
```

# Q2: Named Entity Recognition with SpaCy

```python
import spacy

nlp = spacy.load("en_core_web_sm")

def print_named_entities(sentence):
    doc = nlp(sentence)

    for entity in doc.ents:
        print(
            # Get the entity text 
            f"Text: {entity.text} | "
            # Get the entity label
            f"Label: {entity.label_} | "
            # Get the start and end character positions
            f"Start: {entity.start_char} | "
            f"End: {entity.end_char}"
        )

sentence = (
    " LeBron James as the NBA player of the United States in Laker "
    "and won the NBA Final MVP in 2020."
)

print_named_entities(sentence)
```

# Q3: Scaled Dot-Product Attention 

```python
import numpy as np

def scaled_dot_product_attention(Q, K, V):
    # Compute and scale the query-key dot products.
    key_dim = K.shape[1]
    scores = (Q @ K.T) / np.sqrt(key_dim)

    # Apply softmax separately to each query row.
    shifted_scores = scores - np.max(scores, axis=1, keepdims=True)
    exp_scores = np.exp(shifted_scores)
    attention_weights = exp_scores / np.sum(
        exp_scores, axis=1, keepdims=True
    )

    # Use the attention weights to combine the value vectors.
    output = attention_weights @ V
    return attention_weights, output

Q = np.array([[1, 0, 1, 0], [0, 1, 0, 1]])
K = np.array([[1, 0, 1, 0], [0, 1, 0, 1]])
V = np.array([[1, 2, 3, 4], [5, 6, 7, 8]])

attention_weights, output = scaled_dot_product_attention(Q, K, V)

print("Attention Weights:")
print(np.round(attention_weights, 3))

print("\nFinal Output:")
print(np.round(output, 3))
```
